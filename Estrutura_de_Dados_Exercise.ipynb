```python
class CountryNode:
  def __init__(self, country: str, continent: str, pop_2022: int, area: float):
    self.country = country
    self.continent = continent
    self.pop_2022 = pop_2022
    self.area = area
    self.next = None

  # Calcula a densidade demográfica (hab/km²)
  def density(self) -> float:
    return self.pop_2022 / self.area if self.area > 0 else 0.0

  # Retorna representação em string
  def __str__(self) -> str:
    return (f"País: {self.country:<12} | "
            f"Continente: {self.continent:<15} | "
            f"Pop_2022: {self.pop_2022:>12} | "
            f"Área: {self.area} km² | "
            f"Densidade: {self.density():> 7.2f} hab/km²")

# Lista Simplesmente Encadeada completa para gerenciamento demográfico.
class PopulationLinkedList:
  def __init__(self):
    self.head = None           # Cabeçalho.
    self.tail = None           # Final da lista.
    self.size = 0              # Lista começando do 0

  # ==================== CONSULTAS E ESTADO ====================
  # Verifica se a lista está vazia - O(1)
  def is_empty(self) -> bool:
    return self.head is None

  # Verifica o tamanho da lista
  def __len__(self):
    return self.size

  # Permite iterar sobre a lista usando loops 'for' nativos do Python.
  def __iter__(self):
    current = self.head
    while current:
      yield current
      current = current.next

  # ======================= Métodos de Inserção ======================
  # Adiciona um nó no início da lista - O(1)
  def insert_at_beginning(self, country: str, continent: str, pop_2022: int, area: float):
    new_node = CountryNode(country, continent, pop_2022, area)
    if self.is_empty():
      self.head = new_node
      self.tail = new_node
    else:
      new_node.next = self.head
      self.head = new_node
    self.size += 1

  # Adiciona um nó no final da lista - O(1) com o ponteiro tail.
  def append(self, country: str, continent: str, pop_2022: int, area: float):
    new_node = CountryNode(country, continent, pop_2022, area)
    if self.is_empty():
      self.head = new_node
      self.tail = new_node
    else:
      self.tail.next = new_node
      self.tail = new_node
    self.size += 1

  def insert_at_index(self, index: int, country: str, continent: str, pop_2022: int, area: float):
    if index < 0 or index > self.size:
      raise IndexError("Fora dos limites da lista.")

    if index == 0:
      self.insert_at_beginning(country, continent, pop_2022, area)
      return
    if index == self.size:
      self.append(country, continent, pop_2022, area)
      return

    new_node = CountryNode(country, continent, pop_2022, area)
    current = self.head
    for _ in range(index - 1):
      current = current.next

    new_node.next = current.next
    current.next = new_node
    self.size += 1

  # ========================== Métodos de Remoção =======================
  def remove_first(self) -> CountryNode | None:
    if self.is_empty():
      return None

    removed_node = self.head
    self.head = self.head.next
    self.size -= 1

    if self.is_empty():
      self.tail = None
    return removed_node

  def remove_last(self) -> CountryNode | None:
    if self.is_empty():
      return None
    if self.size == 1:
      return self.remove_first()

    current = self.head
    while current.next != self.tail:
      current = current.next

    removed_node = self.tail
    current.next = None
    self.tail = current
    self.size -= 1
    return removed_node

  def remove_by_name(self, country_name: str) -> bool:
    if self.is_empty():
      return False

    if self.head.country.lower() == country_name.lower():
      self.remove_first()
      return True

    current = self.head
    while current.next and current.next.country.lower() != country_name.lower():
      current = current.next

    if current.next:
      if current.next == self.tail:
        self.tail = current
      current.next = current.next.next
      self.size -= 1
      return True

    return False

  # ============================= BUSCA E ACESSO =========================
  def search(self, country_name: str) -> CountryNode | None:
    for node in self:
      if node.country.lower() == country_name.lower():
        return node
    return None

  def get(self, index: int) -> CountryNode | None:
    if index < 0 or index >= self.size:
      raise IndexError("Índice fora dos limites da lista.")

    current = self.head
    for _ in range(index):
      current = current.next
    return current

  # ===================== ALGORITMOS E UTILITÁRIOS ============================
  def reverse(self):
    previous = None
    current = self.head
    self.tail = self.head

    while current:
      next_node = current.next
      current.next = previous
      previous = current
      current = next_node

    self.head = previous

  def clear(self):
    self.head = None
    self.tail = None
    self.size = 0

  def display(self):
    if self.is_empty():
      print("\n [A lista está vazia]")
      return

    print("\n" + "=" * 98)
    print(f"LISTA DE POPULAÇÃO MUNDIAL (Total de registros: {self.size})")
    print("=" * 98)
    for node in self:
      print(node)
      print("=" * 98 + "\n")

# ===========================================================================
#                        DEMONSTRAÇÃO PRÁTICA DE USO
# ===========================================================================
if __name__ == "__main__":
  lista = PopulationLinkedList()

  lista.append("Brasil", "América do Sul", 215_313_498, 8_515_767.0)
  lista.append("Japão", "Ásia", 125_124_000, 377_975.0)
  lista.insert_at_beginning("China", "Ásia", 1_411_750_000, 9_596_960.0)
  lista.insert_at_index(2, "Alemanha", "Europa", 84_358_845, 357_022.0)

  lista.display()

```

```python
import csv

class Node:
    def __init__(self, country, population):
        self.country = country
        self.population = int(population)
        self.next = None
        self.prev = None

    def __repr__(self):
        return f"{self.country}: {self.population:,}"

class SortedDoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self._length = 0

    def insert(self, country, population):
        """Insere um novo nó mantendo a lista ordenada em ordem crescente de população."""
        new_node = Node(country, population)
        self._length += 1

        # Caso 1: Lista vazia
        if not self.head:
            self.head = new_node
            self.tail = new_node
            print(f"[INSERIR] {new_node} inserido como Head e Tail (Lista estava vazia).")
            return

        # Caso 2: Inserção no início (novo nó é menor que o head)
        if new_node.population < self.head.population:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
            print(f"[INSERIR] {new_node} inserido no início (Head).")
            return

        # Caso 3: Encontrar a posição correta no meio ou no fim
        current = self.head
        while current.next and current.next.population <= new_node.population:
            current = current.next

        # Inserção no final
        if current.next is None:
            current.next = new_node
            new_node.prev = current
            self.tail = new_node
            print(f"[INSERIR] {new_node} inserido no final (Tail).")
        # Inserção no meio
        else:
            new_node.next = current.next
            new_node.prev = current
            current.next.prev = new_node
            current.next = new_node
            print(f"[INSERIR] {new_node} inserido no meio (após {current.country}).")

    def buscar(self, country_name):
        """Busca um país pelo nome e retorna o nó correspondente."""
        print(f"\n[BUSCAR] Procurando por '{country_name}'...")
        current = self.head
        while current:
            if current.country.strip().lower() == country_name.strip().lower():
                print(f" -> Encontrado: {current}")
                return current
            current = current.next
        print(f" -> País '{country_name}' não encontrado.")
        return None

    def deletar(self, country_name):
        """Remove um nó da lista a partir do nome do país."""
        print(f"\n[DELETAR] Tentando remover '{country_name}'...")
        node_to_delete = self.buscar_sem_print(country_name)
        if not node_to_delete:
            print(f" -> Falha: País '{country_name}' não existe na lista.")
            return False

        # Caso 1: O nó a ser deletado é o head
        if node_to_delete == self.head:
            self.head = node_to_delete.next
            if self.head:
                self.head.prev = None
            else:
                self.tail = None

        # Caso 2: O nó a ser deletado é o tail
        elif node_to_delete == self.tail:
            self.tail = node_to_delete.prev
            if self.tail:
                self.tail.next = None

        # Caso 3: O nó está no meio
        else:
            node_to_delete.prev.next = node_to_delete.next
            node_to_delete.next.prev = node_to_delete.prev

        self._length -= 1
        print(f" -> Sucesso: '{country_name}' foi removido da lista.")
        return True

    def buscar_sem_print(self, country_name):
        """Auxiliar interno para buscar nó sem gerar prints repetidos."""
        current = self.head
        while current:
            if current.country.strip().lower() == country_name.strip().lower():
                return current
            current = current.next
        return None

    def print_forward(self, limit=None):
        """Imprime da menor para a maior população."""
        print("\n--- Exibindo Lista (Menor para Maior) ---")
        current = self.head
        count = 0
        if not current:
            print("(Lista vazia)")
            return
        while current and (limit is None or count < limit):
            print(f"  [{count + 1}] {current}")
            current = current.next
            count += 1

    def print_backward(self, limit=None):
        """Imprime da maior para a menor população (usando ponteiros 'prev')."""
        print("\n--- Exibindo Lista (Maior para Menor) ---")
        current = self.tail
        count = 0
        if not current:
            print("(Lista vazia)")
            return
        while current and (limit is None or count < limit):
            print(f"  [{count + 1}] {current}")
            current = current.prev
            count += 1

    def __len__(self):
        return self._length


# --- Bloco de Testes das Funções ---
if __name__ == "__main__":
    lista = SortedDoublyLinkedList()

    print("========================================")
    print("1. TESTANDO INSERÇÕES")
    print("========================================")
    lista.insert("Brasil", 214000000)
    lista.insert("Japão", 125000000)
    lista.insert("Índia", 1408000000)
    lista.insert("Islândia", 375000)

    print("\n========================================")
    print("2. TESTANDO IMPRESSÃO DA LISTA")
    print("========================================")
    lista.print_forward()
    lista.print_backward()

    print("\n========================================")
    print("3. TESTANDO BUSCA")
    print("========================================")
    lista.buscar("Brasil")
    lista.buscar("Alemanha")  # Não existe

    print("\n========================================")
    print("4. TESTANDO DELEÇÃO")
    print("========================================")
    lista.deletar("Japão")     # Elemento do meio
    lista.deletar("Islândia")  # Head (primeiro elemento)
    lista.deletar("França")    # Elemento inexistente

    print("\n========================================")
    print("5. LISTA FINAL APÓS DELEÇÕES")
    print("========================================")
    lista.print_forward()
    print(f"\nTamanho total da lista: {len(lista)} nó(s)")


```

```python
import pandas as pd

#definição do Contêiner do tipo Pilha (Stack - LIFO)

class ConteinerPilhaPopulacao:   #Adiciona os registros populacionais.
  def __init__(self):
    self._itens = []

  def empilhar(self, item):    #Adiciona itens no topo.
    self._itens.append(item)

  def desempilhar(self):           #Remove e retorna o registro no topo.
    if not self.esta_vazia():
      return self._itens.pop()
    raise IndexError("A pilha está vazia.")

  def topo(self):    #Retorna o registro no topo sem remover.
    if not self.esta_vazia():
      return self._itens[-1]
    return None

  def esta_vazia(self):
    return len(self._itens) == 0

  def tamanho(self):
    return len(self._itens)

pilha = ConteinerPilhaPopulacao()
pilha.empilhar("Elemento A")
pilha.empilhar("Elemento B")

print(pilha.topo())         #Output def top elemento B.
print(pilha.desempilhar())  #Output def desempilhar.
print(pilha.tamanho())
print(pilha)


```

```python
#Contêiner Stack (pilhas: last in first out)

class ContenerPilha:
    def __init__(self):
        self._itens = []  # Estrutura interna para armazenar os dados

    def empilhar(self, item):
        """Adiciona um elemento ao topo do contêiner."""
        self._itens.append(item)

    def desempilhar(self):
        """Remove e retorna o elemento do topo."""
        if not self.esta_vazio():
            return self._itens.pop()
        raise IndexError("O contêiner está vazio.")

    def topo(self):
        """Retorna o elemento do topo sem remover."""
        if not self.esta_vazio():
            return self._itens[-1]
        return None

    def esta_vazio(self):
        """Verifica se o contêiner está vazio."""
        return len(self._itens) == 0

    def tamanho(self):
        """Retorna a quantidade de elementos."""
        return len(self._itens)

# Exemplo de uso:
pilha = ContenerPilha()
pilha.empilhar("Elemento A")
pilha.empilhar("Elemento B")

print(pilha.topo())        # Saída: Elemento B
print(pilha.desempilhar()) # Saída: Elemento B
print(pilha.tamanho())     # Saída: 1


print(f'Retorno desempilhar: {(pilha.empilhar("Elemento A"))}, tamanho: {(pilha.empilhar("Elemento B"))}, topo: {(pilha.topo()) }')

```

1  Unidirecional: Cada nó (No) possui apenas uma referência (self.proximo), que aponta exclusivamente para o nó seguinte.

2  Sem navegação reversa: Você consegue percorrer a lista apenas do início (cabeca) em direção ao fim. Não é possível voltar para o nó anterior a partir do nó atual (como acontece em uma Doubly Linked List).
