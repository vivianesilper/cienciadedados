```python
# Interação por linhas head(5)
import pandas as pd

class CountryNode:
    def __init__(self, country, continent, pop_2022, area):
        self.country = country
        self.continent = continent
        self.pop_2022 = pop_2022
        self.area = area
        self.next = None

class PopulationLinkedList:
    def __init__(self):
        self.head = None

    def append(self, country, continent, pop_2022, area):
        new_node = CountryNode(country, continent, pop_2022, area)
        if not self.head:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def search_multiple(self, country_names):

        targets = [c.lower() for c in country_names]
        found_nodes = []
        current = self.head

        while current:
            if current.country.lower() in targets:
              found_nodes.append(current)

            current = current.next  # <-- CORRIGIDO AQUI (antes estava next.current)

        return found_nodes
```

```python
import pandas as pd

# 1. Carrega o CSV completo na Linked List
df = pd.read_csv(r'/content/world_population.csv', sep=';')
country_list = PopulationLinkedList()

for _, row in df.iterrows():
    country_list.append(
        country=row['Country/Territory'],
        continent=row['Continent'],
        pop_2022=row['2022 Population'],
        area=row['Area (km²)']
    )

# 2. Define a lista com os países que você quer buscar
paises_para_buscar = ['Brazil', 'Afghanistan', 'Argentina', 'Canada', 'Japan']

# 3. Executa a busca
resultados = country_list.search_multiple(paises_para_buscar)

# 4. Imprime os resultados encontrados
for no in resultados:
    print(f"Encontrado: {no.country} | Continente: {no.continent} | Pop. 2022: {no.pop_2022} | Área: {no.area}")
```
