Carrega o `world_population.csv` (separador `;`, UTF-8) para um DataFrame.

```python
import pandas as pd
from pathlib import Path

CAMINHO_CSV = Path('data') / 'world_population.csv'
```

```python
df = pd.read_csv(CAMINHO_CSV, sep=';', encoding='utf-8')
print(f'{df.shape[0]} linhas x {df.shape[1]} colunas')
df.head()
```
