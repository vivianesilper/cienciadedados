**Notes**

**Batch Size** é um termo usado em aprendizado de máquina e refere-se ao número de exemplos de treinamento usados em uma iteração.

**Bath mode** onde o tamanho do lote é igual ao conjunto de dados total, tornando os valores de iteração e épocas equivalentes.

**mini-batch mode** onde o tamanho do lote é maior do que um, mas menor que o tamanho total do conjunto de dados, um número que pode ser dividido no tamanho total do conjunto de dados.

**stochastic mode**
onde o tamanho do lote é igual a um. O gradiente e os parâmetros da rede neural são atualizados após cada amostra.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import confusion_matrix
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten, Embedding
from google.colab import files

from keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
```

```python
files.upload()
```

```python
spam = pd.read_csv("spam.csv")
spam.head()
```

```python
labelencoder = LabelEncoder()    #hot encode dos dados, transformando dados (words) dados em númericos.
y = labelencoder.fit_transform(spam['Category'])
print(y)
```

```python
mensagem = spam['Message'].values
X_train, X_test, y_train, y_test = train_test_split(mensagem, y, test_size=0.3)
print(X_train)
```

```python
token = Tokenizer(num_words=1000)
token.fit_on_texts(X_train)

X_train = token.texts_to_sequences(X_train)
X_test = token.texts_to_sequences(X_test)
```

**RNA 2**

```python
modelo = Sequential()
modelo.add(Embedding(input_dim=len(token.word_index), output_dim=50,input_length=500))
modelo.add(Flatten())

modelo.add(Dense(units=10,activation="relu"))
modelo.add(Dropout(0.1))
modelo.add(Dense(units=1,activation="sigmoid"))
```

```python
modelo.compile(loss="mean_squared_error", optimizer="adam",metrics=["accuracy"])
modelo.summary()
```

```python
modelo.fit(X_train, y_train, epochs=5, batch_size=10, verbose=True, validation_data=(X_test, y_test))
```

```python
loss, accuracy = modelo.evaluate(X_test,y_test)
print("Loss: ", loss)
print("Acurácia: ", accuracy)
```
