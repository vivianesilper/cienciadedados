**Método Stemming**: extrair a raiz de cada palavra (radical)

*   **Porter**: Mais comum e mais antigo.
*   **Snowball**: Melhorando com relação do Porter. Melhor performance computacional.
*   **Lancaster**: Mais agressivo. Resultado as vezes não intuítivo.



```python
#Porter
stemmer = PorterStemmer()
stem1 = [stemmer.stem(word) for word in palavras_sem_pontuacao2]
print(palavras_sem_pontuacao2)
print(stem1)
```

```python
print(len(stem1))
```

```python
#Snowball
stemmer2 = SnowballStemmer("portuguese")
stem2 = [stemmer.stem(word) for word in palavras_sem_pontuacao2]
print(palavras_sem_pontuacao2)
print(stem2)
```

```python
#LancasterStemmer
stemmer3 = LancasterStemmer()
stem3 = [stemmer3.stem(word) for word in palavras_sem_pontuacao2]
print(palavras_sem_pontuacao2)
print(stem3)
```

**Camada de Ativação Multilayer (Perceptron)**

Camadas da Redes Neurais

*   Camada de entrada(input)
Classificação:atributos.

*   Camadas ocultas: Hiperparâmetros (sinapses)
Quantidade de Neurônios na camada com seus hiperparâmetros
sinapses: onde ficam configurados os pesos e a relação de comunicação entre cada neurônio.

*   Camada saída(output)



Backpropagation (configuração da camada)

A flexibilidade potencial de uma Rede Neural Artificial está proporcionalmente relacionada á quantidade de camadas ocultas e seus tamanhos. Isso quer dizer quanto mais camadas ocultas ela tiver e quanto maior a quantidade de neurônios possuir, maior será a capacidade de aprendizagem e maiores as chances de separar os dados corretamente.

Ajuste dos pesos.

Funções de ativação das sinapses neurais



*   Threshold: ativação binária (True and False)

*   Sigmoid:somas ponderadas, comumente utilizada na regressão logistica classificação, prever probabilidade, açoes, tendências e categorias.

*   Relu (Rectified Lenar Unit):

*   Ranh (Hyperbolic tangent activation function)




Regularização

*   Técnicas de evitar ou diminuir o overfitting .
*   Um "termo" de regularização é adicionado á medida que os pesos são aprendidos
*   O termo L é a soma dos pesos
*   O termo L2 é a soma do quadrado dos pesos.

**Regularização**

L1: soma dos pesos

* executa a seleção de atributos inteiros vão para 0.
* Computacionalmente ineficiente
* Sáda esparsa.

L2:soma do quadrado dos pesos

* Todos os atributos permanecem, são apenas ponderados.
* Computacionalmete eficiente
* Saída densa.

**Regularização-Dropout-Deep Learning**

Pocesso de modificação da rede neural, começa por eliminar aleatoriamente alguns neurônios ocultos na rede permanecendo os neurônios de entrada e saída intocados, minimizando o super ajustes.
Você ira definir seu ajuste no parâmetro.

**Regularização Early stopping**
O treinamento ira diminuir a acurracia para verificar o grau de concordância dos dados, para o treinamento quando a acurrancia começa a diminuir drasticamente.
